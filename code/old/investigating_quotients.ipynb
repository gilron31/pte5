```python
import utils
import sympy as sp
import itertools
from collections import defaultdict
```

```python
x, y = sp.var("x, y")
p = 7


pol = sp.poly(x**2)
[pol(x) % p for x in range(p)]
```

```python
def get_all_values(pol, p, n_vars):
    rv = defaultdict(list)
    for v in itertools.product(range(p), repeat=n_vars):
        rv[pol(*v) % p].append(v)
    return rv


def get_all_values_rat(pol_num, pol_den, p, n_vars):
    rv = defaultdict(list)
    for v in itertools.product(range(p), repeat=n_vars):
        num = pol_num(*v) % p
        den = pol_den(*v) % p
        rv[(pow(den, -1, p) * num % p) if den != 0 else "inf"].append(v)
    return rv


# get_all_values(sp.poly(x**2 + y**2), 7, 2)
onepoint_q_val = get_all_values_rat(
    sp.poly(x * y) ** 2, sp.poly(x**2 + y**2) ** 2, 7, 2
)
# onepoint_q_val
```

```python
p = 5
onepoint_q_val = get_all_values_rat(
    sp.poly(x * y) ** 2, sp.poly(x**2 + y**2) ** 2, p, 2
)
two_point_q_val = defaultdict(list)
for k, v in onepoint_q_val.items():
    for k2, v2 in onepoint_q_val.items():
        two_point_q_val[(k + k2) % p if ("inf" not in [k, k2]) else "inf"] = v + v2
two_point_q_val.keys()
```
