```python
def verify_skew_normal_symbolic(cs):
    x = sp.var("x")
    p = sp.poly(x)
    for c in cs:
        p = p**2 - c
    iroots = [r for r in p.all_roots() if r.is_Integer]
    is_gem = len(iroots) == 2 ** len(cs)
    print(f"{p=} \n{iroots=} \n{is_gem=}")
    return p, iroots, is_gem


p, iroots, is_gem = verify_skew_normal_symbolic([1, 2, 3])
p, iroots, is_gem = verify_skew_normal_symbolic([85, 4176, 2880**2])
```

```python
def compute_square_roots(l):
    rv = []
    for x in l:
        if x <= 0:
            return None
        r = sp.integer_nthroot(x, 2)
        if r[1]:
            rv.append(r[0])
        else:
            return None
    return rv


def verify_skew_normal_squareish(cs):
    iroots = [cs[-1]]
    iroots = compute_square_roots(iroots)

    is_gem = True
    for c in cs[-2::-1] + [0]:
        # print(c, iroots)
        if iroots:
            iroots = [r for r_ in iroots for r in [c - r_, c + r_]]
        else:
            is_gem = False
            break
        if c != 0:
            iroots = compute_square_roots(iroots)
    print(f"{iroots=} \n{is_gem=}")
    return iroots, is_gem


verify_skew_normal_squareish([1, 2, 81])
verify_skew_normal_squareish([85, 4176, 2880**2])
```
